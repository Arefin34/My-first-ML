```python
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
documents = [
    "A car is a vehicle used for transportation.",
    "Python is a popular programming language.",
    "You can purchase an automobile from a dealership.",
    "Machine learning allows computers to learn from data.",
    "A bicycle is a human-powered vehicle.",
    "Engineering university has good engineers"
]
```

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
document_embeddings = model.encode(documents, normalize_embeddings=True)

def search(query, top_k = 3):
  query_embedding = model.encode(query, normalize_embeddings=True)


  score = document_embeddings @ query_embedding

  ranked_indice = np.argsort(score)[: : -1][: top_k]



  for index in ranked_indice:
    print(f"Score: {score[index]:.3f}")
    print(documents[index])
    print()
```

```python
search("Where can i find engineeers?")
```
